## <u>AFP Factuel</u>

### **Description :**
C’est la source d'autorité en France. Le site est principalement statique, ce qui rend Scrapy idéal pour extraire des milliers d'articles.

### **Site web :** https://factuel.afp.com/

### **Methode :** scrapping

### **Type de données :** Articles de debunking, texte d'analyse et images (souvent des captures d'écran de réseaux sociaux avec un tampon "Faux").

### **Format :** JSON ou CSV.

### **Langue :** Français.

Qualité des labels : Maximale (Vrai/Faux/Partiellement Faux).

---

## <u>Reddit - via PRAW</u>

### **Description :**
Reddit est une mine d'or pour la désinformation "spontanée" (memes, captures d'écran modifiées). L'API officielle est très robuste.

### **Méthode :** API REST

### **Type de données :** Titres de posts, images (URL directes vers i.redd.it ou Imgur) et métadonnées (nombre d'upvotes, commentaires).

### **Format :** JSON.

### **Langue :** Multilingue (cibler les subreddits comme r/fakenews ou r/PropagandaPosters).

Qualité des labels : Moyenne. Tu devras utiliser les "flairs" (étiquettes de posts) ou le score de confiance de la communauté comme indicateur.

---

## <u>Les Observateurs - France 24</u>

### **Description:**
Une émission et un site dédiés à la vérification d'images et de vidéos envoyées par les internautes.

### **Méthode :** Selenium

### **Type de données :** Galerie d'images avant/après (image originale vs image détournée) et explications textuelles.

### **Format :** CSV (pour le texte) + Dossier local (pour les images).

### **Langue :** Français.

Qualité des labels : Maximale.

---

## <u>Google Fact Check Explorer</u>

### **Description**
Contrairement aux autres sources qui sont des producteurs de contenu, celle-ci est un agrégateur. Elle utilise le standard ClaimReview (un balisage schema.org utilisé par les fact-checkers du monde entier).

### **Méthode :** API REST

### **Type de données :** La "Claim" (l'affirmation fausse), l'auteur de l'affirmation, le verdict (vrai/faux), et surtout l'URL vers l'article source qui contient souvent l'image originale.

### **Format :** JSON.

### **Langue :** Multilingue (filtrer par langue, ex: languageCode=fr).

Qualité des labels : Maximale. Les données proviennent d'organisations certifiées par l'IFCN (International Fact-Checking Network).
